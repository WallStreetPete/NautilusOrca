# First backtest

End-to-end: ingest NVDA from yfinance → run the SMA cross strategy → tearsheet.

```python
from datetime import date, timedelta
from blackorca.data.sources.yfinance import YFinanceSource
from blackorca.data.catalog import Catalog
from blackorca.data.ingest import ingest_bars
from blackorca.backtest.runner import run_backtest
from blackorca.backtest.analyzer import render_html_tearsheet
from blackorca.risk.limits import RiskLimits
from blackorca.strategies.examples.sma_cross import SmaCross

cat = Catalog()
src = YFinanceSource()
end = date.today(); start = end - timedelta(days=365*4)
ingest_bars(src, cat, ['NVDA'], start, end)

relaxed = RiskLimits(per_order_max_notional=1e9, max_position_pct=0.5)
strat = SmaCross(symbol='NVDA', fast=10, slow=30, target_weight=0.2)
res = run_backtest(strat, symbols=['NVDA'], start=start, end=end, catalog=cat, risk_limits=relaxed)
print(res.metrics)
render_html_tearsheet(res.metrics, res.equity_curve, title='SMA cross NVDA', out_path='data/reports/01_first_backtest.html')
```
