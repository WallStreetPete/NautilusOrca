# Event study template

Market-adjusted CAR/AAR around earnings dates.

```python
from datetime import date, datetime, timezone
import polars as pl
from blackorca.data.catalog import Catalog
from blackorca.research.event_study import run_event_study

cat = Catalog()
prices = cat.read_bars('NVDA', date(2020,1,1), date.today())
events = pl.DataFrame({'symbol': ['NVDA']*4, 'event_date': [datetime(y,m,15,tzinfo=timezone.utc) for y,m in [(2023,2),(2023,5),(2023,8),(2023,11)]]})
res = run_event_study(prices, events, pre_window=3, post_window=10)
print(res.n_events)
res.aar
```
