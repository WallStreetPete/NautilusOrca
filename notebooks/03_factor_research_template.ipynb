# Factor research template

```python
from datetime import date
import polars as pl
from blackorca.data.catalog import Catalog
from blackorca.research.factor_research import run_factor_study
from blackorca.research.ic_analysis import compute_ic_decay
from blackorca.universe.semis import symbols as semi_symbols

cat = Catalog()
syms = semi_symbols()
prices = cat.read_bars(syms, date(2020,1,1), date.today())
factor = prices.sort(['symbol','as_of']).with_columns(value=-pl.col('close').pct_change().shift(1).over('symbol')).drop_nulls('value').select(['symbol','as_of','value'])
print(compute_ic_decay(factor, prices))
run_factor_study(factor, prices, n_buckets=5).metrics
```
