# Agent research loop demo

Run a bounded autonomous research loop on a semis-focused universe context. Costs ~$0.05 per iteration on Sonnet 4.6.

```python
from blackorca.agents.graphs.research_loop import run_research_loop

state = run_research_loop(
    universe_context=(
        'Universe: US semis (~25 names). Recent: TSMC monthly revenue +60% YoY. We have daily bars '
        'and a supplier dependency graph in the catalog.'
    ),
    max_iterations=1,
    fallback_strategy='sma_cross',
    fallback_symbol='NVDA',
)
print('cost:', round(state.total_cost_usd, 4))
print('recommendation:', state.final_recommendation)
print('hypothesis:', state.hypothesis.statement if state.hypothesis else None)
print('analysis recommendation:', state.analysis.recommendation if state.analysis else None)
```
